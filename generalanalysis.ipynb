```python
### Of all matching notices for each prefecture, what percentage are goods, constructions, or services?



hits_pivot = counts.pivot(
       index="prefecture_en",
       columns="category_en",
       values="search_hits",
   )

category_ratios = hits_pivot.div(
       hits_pivot.sum(axis=1),
       axis=0,
   )

category_ratios

category_percentages = category_ratios * 100

category_percentages.round(1)
```

```python
ax = category_percentages.plot(
       kind="bar",
       stacked=True,
       figsize=(12, 6),
       title="Procurement Category Share by Prefecture",
   )

ax.set_xlabel("Prefecture")
ax.set_ylabel("Share of Search Hits (%)")
ax.legend(title="Category")
```

```python
organization_coverage = (
       notices.dropna(subset=["organization_name"])
       .groupby(["prefecture_name", "organization_name"])
       .agg(
           sample_notices=("key", "count"),
           categories=("category", "nunique"),
       )
       .reset_index()
   )

all_category_organizations = organization_coverage[
       organization_coverage["categories"] == 3
   ].sort_values(
       ["prefecture_name", "sample_notices"],
       ascending=[True, False],
   )

all_category_organizations.head(30)
```

```python
top_organizations = (
       notices.dropna(subset=["organization_name"])
       .groupby(
           ["prefecture_name", "category", "organization_name"],
           as_index=False,
       )
       .size()
       .rename(columns={"size": "sample_notices"})
       .sort_values(
           ["prefecture_name", "category", "sample_notices"],
           ascending=[True, True, False],
       )
   )

top_organizations.groupby(
       ["prefecture_name", "category"]
   ).head(10)
```

```python
 organization_counts = (
       notices.dropna(subset=["organization_name"])
       .groupby(
           ["prefecture_name", "category", "organization_name"],
           as_index=False,
       )
       .size()
       .rename(columns={"size": "sample_notices"})
   )
```

```python
import matplotlib.pyplot as plt

   # A Japanese font commonly available on macOS
plt.rcParams["font.family"] = "Hiragino Sans"

category_order = ["物品", "工事", "役務"]

category_english = {
       "物品": "Goods",
       "工事": "Construction",
       "役務": "Services",
   }

def plot_top_organizations(prefecture, top_n=10):
       prefecture_data = organization_counts[
           organization_counts["prefecture_name"] == prefecture
       ]

       fig, axes = plt.subplots(
           nrows=1,
           ncols=3,
           figsize=(22, 8),
       )

       for ax, category in zip(axes, category_order):
           top = (
               prefecture_data[
                   prefecture_data["category"] == category
               ]
               .nlargest(top_n, "sample_notices")
               .sort_values("sample_notices")
           )

           ax.barh(
               top["organization_name"],
               top["sample_notices"],
           )

           ax.set_title(category_english[category])
           ax.set_xlabel("Notices in collected sample")
           ax.set_ylabel("Organization")

       fig.suptitle(
           f"Top Issuing Organizations in {prefecture}",
           fontsize=16,
       )


       plt.tight_layout()
       plt.show()
```

```python
notices["issue_date"] = pd.to_datetime(
       notices["cft_issue_date"],
       errors="coerce",
       utc=True,
   )

notices["issue_date_jst"] = (
       notices["issue_date"]
       .dt.tz_convert("Asia/Tokyo")
   )

notices["issue_month"] = (
       notices["issue_date_jst"]
       .dt.strftime("%Y-%m")
   )
```

```python
monthly_sample = (
       notices.groupby(["issue_month", "category"])
       .size()
       .reset_index(name="sample_notices")
   )

monthly_sample.head()
```

```python
notices["issue_year"] = (
       notices["issue_date_jst"]
       .dt.year
   )

yearly_sample = (
       notices.dropna(subset=["issue_year"])
       .groupby(
           ["prefecture_name", "issue_year", "category"],
           as_index=False,
       )
       .size()
       .rename(columns={"size": "sample_notices"})
   )

yearly_sample.head()
```

```python
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Hiragino Sans"

prefectures = sorted(
       yearly_sample["prefecture_name"].unique()
   )

category_order = ["物品", "工事", "役務"]

fig, axes = plt.subplots(
       nrows=2,
       ncols=3,
       figsize=(20, 11),
       sharex=True,
   )

axes = axes.flatten()

for ax, prefecture in zip(axes, prefectures):
       prefecture_data = yearly_sample[
           yearly_sample["prefecture_name"] == prefecture
       ]

       annual_pivot = prefecture_data.pivot(
           index="issue_year",
           columns="category",
           values="sample_notices",
       ).fillna(0)

       # Keep categories in the same order and colors across every chart.
       annual_pivot = annual_pivot.reindex(
           columns=category_order,
           fill_value=0,
       )

       annual_pivot.plot(
           ax=ax,
           marker="o",
       )

       ax.set_title(prefecture)
       ax.set_xlabel("Issue year")
       ax.set_ylabel("Notices in collected sample")
       ax.grid(alpha=0.3)
       ax.legend(title="Category")

   # These lines must be outside the loop.
fig.suptitle(
       "Annual Procurement Notice Dates in the Collected Sample",
       fontsize=18,
   )

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
annual_counts = pd.read_csv(
       "one_off_scripts/annual_counts.csv",
       dtype={"lg_code": str},
   )

annual_counts = annual_counts.sort_values(
       ["prefecture_ja", "category_ja", "year"]
   ).copy()

annual_counts["yoy_growth_pct"] = (
       annual_counts
       .groupby(
           ["prefecture_ja", "category_ja"]
       )["search_hits"]
       .pct_change(fill_method=None)
       .mul(100)
   )
```

```python
annual_counts[
       [
           "year",
           "prefecture_ja",
           "category_ja",
           "search_hits",
           "yoy_growth_pct",
       ]
   ].round(1)
```

```python
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Hiragino Sans"

prefectures = sorted(
       annual_counts["prefecture_ja"].unique()
   )

category_order = ["物品", "工事", "役務"]

category_colors = {
       "物品": "tab:blue",
       "工事": "tab:orange",
       "役務": "tab:green",
   }

fig, axes = plt.subplots(
       nrows=2,
       ncols=3,
       figsize=(20, 11),
       sharex=True,
   )

axes = axes.flatten()

for ax, prefecture in zip(axes, prefectures):
       prefecture_data = annual_counts[
           annual_counts["prefecture_ja"] == prefecture
       ]

       for category in category_order:
           category_data = prefecture_data[
               prefecture_data["category_ja"] == category
           ].dropna(subset=["yoy_growth_pct"])

           ax.plot(
               category_data["year"],
               category_data["yoy_growth_pct"],
               marker="o",
               label=category,
               color=category_colors[category],
           )

       # Zero separates growth from decline.
       ax.axhline(
           0,
           color="black",
           linewidth=1,
           alpha=0.5,
       )

       ax.set_title(prefecture)
       ax.set_xlabel("Year")
       ax.set_ylabel("Year-over-year change (%)")
       ax.grid(alpha=0.3)
       ax.legend(title="Category")

fig.suptitle(
       "Year-over-Year Change in KKJ Procurement Notices",
       fontsize=18,
   )

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
population = pd.read_csv(
       "reference_data/prefecture_population.csv",
       dtype={"lg_code": str},
   )

area = pd.read_csv(
       "reference_data/prefecture_area.csv",
       dtype={"lg_code": str},
   )
```

```python
annual_enriched = annual_counts.merge(
       population[
           ["year", "lg_code", "population"]
       ],
       on=["year", "lg_code"],
       how="left",
       validate="many_to_one",
   )

annual_enriched["notices_per_100k"] = (
       annual_enriched["search_hits"]
       / annual_enriched["population"]
       * 100_000
   )
```

```python
annual_enriched[
       [
           "year",
           "prefecture_en",
           "category_en",
           "search_hits",
           "population",
           "notices_per_100k",
       ]
   ].round(2)
```

```python
rates_2025 = annual_enriched[
       annual_enriched["year"] == 2025
   ]

rates_2025_pivot = rates_2025.pivot(
       index="prefecture_en",
       columns="category_en",
       values="notices_per_100k",
   ).reindex(
       columns=["goods", "construction", "services"],
   )

ax = rates_2025_pivot.plot.bar(
       figsize=(13, 7),
   )

ax.set_title(
       "Procurement Notices per 100,000 Residents, 2025"
   )
ax.set_xlabel("Prefecture")
ax.set_ylabel("Notices per 100,000 residents")
ax.legend(title="Category")
ax.grid(axis="y", alpha=0.3)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(
       2,
       3,
       figsize=(20, 11),
       sharex=True,
   )

axes = axes.flatten()

for ax, prefecture in zip(
       axes,
       sorted(annual_enriched["prefecture_en"].unique()),
   ):
       prefecture_data = annual_enriched[
           annual_enriched["prefecture_en"] == prefecture
       ]

       pivot = prefecture_data.pivot(
           index="year",
           columns="category_en",
           values="notices_per_100k",
       ).reindex(
           columns=["goods", "construction", "services"],
       )

       pivot.plot(ax=ax, marker="o")

       ax.set_title(prefecture)
       ax.set_xlabel("Year")
       ax.set_ylabel("Notices per 100,000 residents")
       ax.grid(alpha=0.3)
       ax.legend(title="Category")

fig.suptitle(
       "Annual Procurement Notices per 100,000 Residents",
       fontsize=18,
   )

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
area = pd.read_csv(
       "reference_data/prefecture_area.csv",
       dtype={"lg_code": str},
   )

annual_geo = annual_enriched.merge(
       area[["lg_code", "area_km2"]],
       on="lg_code",
       how="left",
       validate="many_to_one",
   )

annual_geo["notices_per_1000_km2"] = (
       annual_geo["search_hits"]
       / annual_geo["area_km2"]
       * 1_000
   )
```

```python
construction_area_2025 = (
       annual_geo[
           (annual_geo["year"] == 2025)
           & (annual_geo["category_en"] == "construction")
       ]
       .sort_values("notices_per_1000_km2")
   )

ax = construction_area_2025.plot.barh(
       x="prefecture_en",
       y="notices_per_1000_km2",
       figsize=(10, 6),
       legend=False,
   )

ax.set_title(
       "Construction Notices per 1,000 km², 2025"
   )
ax.set_xlabel("Construction notices per 1,000 km²")
ax.set_ylabel("Prefecture")
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```
