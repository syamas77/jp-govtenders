```python
### Search Hits by category 

category_totals = (
    counts.groupby("category_en", as_index=False)["search_hits"]
    .sum()
    .sort_values("search_hits", ascending=False)
)

category_totals
```

```python
category_totals.plot.bar(
       x="category_en",
       y="search_hits",
       legend=False,
       title="Procurement Search Hits by Category",
)
```

```python
counts.pivot(
       index="prefecture_en",
       columns="category_en",
       values="search_hits",
   ).plot.bar(
       figsize=(12, 6),
       title="Procurement Search Hits by Prefecture and Category",
   )
```

```python
### Of all matching notices for each prefecture, what percentage are goods, constructions, or services?



hits_pivot = counts.pivot(
       index="prefecture_en",
       columns="category_en",
       values="search_hits",
   )

category_ratios = hits_pivot.div(
       hits_pivot.sum(axis=1),
       axis=0,
   )

category_ratios

category_percentages = category_ratios * 100

category_percentages.round(1)
```

```python
ax = category_percentages.plot(
       kind="bar",
       stacked=True,
       figsize=(12, 6),
       title="Procurement Category Share by Prefecture",
   )

ax.set_xlabel("Prefecture")
ax.set_ylabel("Share of Search Hits (%)")
ax.legend(title="Category")
```

```python
prefecture_totals = hits_pivot.sum(axis=1).sort_values(ascending=False)

prefecture_totals
```

```python
organization_coverage = (
       notices.dropna(subset=["organization_name"])
       .groupby(["prefecture_name", "organization_name"])
       .agg(
           sample_notices=("key", "count"),
           categories=("category", "nunique"),
       )
       .reset_index()
   )

all_category_organizations = organization_coverage[
       organization_coverage["categories"] == 3
   ].sort_values(
       ["prefecture_name", "sample_notices"],
       ascending=[True, False],
   )

all_category_organizations.head(30)
```

```python
top_organizations = (
       notices.dropna(subset=["organization_name"])
       .groupby(
           ["prefecture_name", "category", "organization_name"],
           as_index=False,
       )
       .size()
       .rename(columns={"size": "sample_notices"})
       .sort_values(
           ["prefecture_name", "category", "sample_notices"],
           ascending=[True, True, False],
       )
   )

top_organizations.groupby(
       ["prefecture_name", "category"]
   ).head(10)
```

```python
 organization_counts = (
       notices.dropna(subset=["organization_name"])
       .groupby(
           ["prefecture_name", "category", "organization_name"],
           as_index=False,
       )
       .size()
       .rename(columns={"size": "sample_notices"})
   )
```

```python
import matplotlib.pyplot as plt

   # A Japanese font commonly available on macOS
plt.rcParams["font.family"] = "Hiragino Sans"

category_order = ["物品", "工事", "役務"]

category_english = {
       "物品": "Goods",
       "工事": "Construction",
       "役務": "Services",
   }

def plot_top_organizations(prefecture, top_n=10):
       prefecture_data = organization_counts[
           organization_counts["prefecture_name"] == prefecture
       ]

       fig, axes = plt.subplots(
           nrows=1,
           ncols=3,
           figsize=(22, 8),
       )

       for ax, category in zip(axes, category_order):
           top = (
               prefecture_data[
                   prefecture_data["category"] == category
               ]
               .nlargest(top_n, "sample_notices")
               .sort_values("sample_notices")
           )

           ax.barh(
               top["organization_name"],
               top["sample_notices"],
           )

           ax.set_title(category_english[category])
           ax.set_xlabel("Notices in collected sample")
           ax.set_ylabel("Organization")

       fig.suptitle(
           f"Top Issuing Organizations in {prefecture}",
           fontsize=16,
       )


        plt.tight_layout()
        plt.show()
```
